```python
contact_book = {}
for i in range(5):
    v = input("Enter contact number: ")
    k = input("Enter contact name: ")
    contact_book[k] = v
print('updated contact book:', contact_book)

```

```python
def search_contact(contact_book):
    contact_number = input("Enter contact number to search: ")
    for name, number in contact_book.items():
        if number == contact_number:
            return f'Contact name: {name}, Contact number: {number}'
        else:
            return f'Contact number {contact_number} not found in contact book'
search_contact(contact_book)


print(search_contact(contact_book))
```

```python
def update_contact_phone_number(contact_book):
    contact_number = input("Enter contact number to update: ")
    if contact_number in contact_book:
        new_contact_number = input("Enter new contact number: ")
        contact_book[new_contact_number] = contact_book.pop(contact_number)
        return "Contact number updated successfully"
    else:
        return "Contact not found"
print(update_contact_phone_number(contact_book))
```

```python
print('updated contact book:', contact_book)
```

```python
def delte_contact(contact_book):
    contact_number = input("Enter contact number to delete: ")
    if contact_number in contact_book:
        del contact_book[contact_number]
        return "Contact deleted successfully"
    else:
        return "Contact not found"
print(delte_contact(contact_book))
```

```python
print('updated contact book:', contact_book)
```

```python
S = { "John": 85,
     "Mary": 92, 
     "Paul": 78, 
     "Anna": 89, 
     "Steve": 95 
    
}
```

```python
def calculate_average_score(S):
    total_score = sum(S.values())
    average_score = total_score / len(S)
    return average_score
calculate_average_score(S)
```

```python
def find_highest_score(S):
    highest_score = max(S.values())
    highest_scorer = [name for name, score in S.items() if score == highest_score]
    return highest_scorer, highest_score
find_highest_score(S)
```

```python
def count_word_frequency(text):
    word_frequency = {}
    words = text.split()
    for w in words:
        word_frequency[w] = word_frequency.get(w,0) + 1
    return word_frequency
count_word_frequency("This is a sample text. This text is for testing the word frequency count function. The function should count the frequency of each word in the given text.")
```

```python
import random

# ---------------------------------------------------------
# 1. Data setup
# ---------------------------------------------------------
courses = {
    'math': {'dom': random.randint(0, 100),
             'menghour': random.randint(0, 100),
             'dara': random.randint(0, 100)},
    'physics': {'pitu': random.randint(0, 100),
                'raboth': random.randint(0, 100),
                'lim': random.randint(0, 100)},
    'chemistry': {'alice': random.randint(0, 100),
                  'david': random.randint(0, 100),
                  'sokhom': random.randint(0, 100)},
}

random.seed(42) # remove this line if you want a different random roster each run



def average_grade(course_name):

    grades = courses[course_name].values()
    return sum(grades) / len(grades)


def highest_grade(course_name):
    """Return the name of the student with the highest grade in a course."""
    students = courses[course_name]
    return max(students, key=students.get)


def passing_students(course_name):
    """Return a set of names of students who passed (grade >= 50)."""
    return {name for name, grade in courses[course_name].items() if grade >= 50}

def failed_students(course_name):
    """Return a set of names of students who failed (grade < 50)."""
    return {name for name, grade in courses[course_name].items() if grade < 50}


print("Course data:")
for course, students in courses.items():
    print(f"  {course}: {students}")
print()

print(f"Average grade in Math: {average_grade('math')}")
print(f"{highest_grade('physics')} has the highest grade in Physics.")

pass_chem = passing_students('chemistry')
print(f"Passing students in Chemistry: {pass_chem if pass_chem else 'None'}")

fail_math = failed_students('math')
print(f"Failed students in Math: {fail_math if fail_math else 'None'}")
```
