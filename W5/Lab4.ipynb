**Exercise 1: Count all gifts**

```python
box = ["book", ["pen", "pencil"],
       [["toy"], "watch"]]
def count_gifts(box):
    total = 0
    for item in box:
        if isinstance(item, list):
            total += count_gifts(item)
        else:
            total += 1
    return total
```

```python
print("count_gifts([]):", count_gifts([]))
print("count_gifts(['book']):", count_gifts(["book"]))
print("count_gifts(box):", count_gifts(box))
print("count_gifts(nested):", count_gifts([[], ["ball"], [["kite", "ball"]]]))
print("count_gifts([[['ring']]]):", count_gifts([[["ring"]]]))
print("extra count:", count_gifts([["a", ["b"]], "c", [[]]]))
```

**Exercise 2: Search for a gift**

```python
def contains_gift(box, target):
    for item in box:
        if isinstance(item, list):
            if contains_gift(item, target):
                return True
        elif item == target:
            return True
    return False
```

```python
print("toy:", contains_gift(box, "toy"))
print("watch:", contains_gift(box, "watch"))
print("phone:", contains_gift(box, "phone"))
print("Toy:", contains_gift(box, "Toy"))
print("book:", contains_gift([], "book"))
print("[[], ['pen']]:", contains_gift([[], ["pen"]], "pen"))
print("extra:", contains_gift([["a"], ["b"]], "b"))
```

**Exercise 3: Palindrome numbers**

```python
def isPalindromeNum(n):
    digits = str(n)
    def check(left, right):
        if left >= right:
            return True
        if digits[left] != digits[right]:
            return False
        return check(left + 1, right - 1)
    return check(0, len(digits) - 1)
```

```python
for n in [0, 7, 11, 123454321, 1357531, 168861]:
    print(f"isPalindromeNum({n}):", isPalindromeNum(n))
for n in [10, 112221]:
    print(f"isPalindromeNum({n}):", isPalindromeNum(n))
print("extra 12341:", isPalindromeNum(12341))
# What stops it: check() stops when left >= right, or when two digits don't match.
# What shrinks: the gap between left and right shrinks by 2 each call.
# Why early False fails: returning False before comparing would reject valid
# palindromes without actually checking their digits.
# Extra test: isPalindromeNum(1) checks a single digit (left == right at start),
# catching a bug if the base case used > instead of >=.
```

**Challenge Exercise 1: Unpack all gifts**

```python
def unpack(box):
    result = []
    for item in box:
        if isinstance(item, list):
            result += unpack(item)
        else:
            result.append(item)
    return result
print(unpack(box))
print(unpack([["ball", "ball"]]))
print(unpack([]))
```

**Challenge Exercise 2: Find the deepest box**

```python
def box_depth(box):
    deepest = 0
    for item in box:
        if isinstance(item, list):
            deepest = max(deepest, box_depth(item))
    return 1 + deepest
print(box_depth([]))
print(box_depth(["book", "pen"]))
print(box_depth([["book"]]))
print(box_depth(["book", [["toy"]]]))
print(box_depth([[]]))
```
