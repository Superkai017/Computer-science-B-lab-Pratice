**Exercise 1: Count all gifts**

```python
box = ["book", ["pen", "pencil"],
       [["toy"], "watch"]]
def count_gifts(box):
    total = 0
    for item in box:
        if isinstance(item, list):
            total += count_gifts(item)
        else:
            total += 1
    return total
```

```python
print("count_gifts([]):", count_gifts([]))
print("count_gifts(['book']):", count_gifts(["book"]))
print("count_gifts(box):", count_gifts(box))
print("count_gifts(nested):", count_gifts([[], ["ball"], [["kite", "ball"]]]))
print("count_gifts([[['ring']]]):", count_gifts([[["ring"]]]))
print("extra count:", count_gifts([["a", ["b"]], "c", [[]]]))
```

**Exercise 2: Search for a gift**

```python
def contains_gift(box, target):
    for item in box:
        if isinstance(item, list):
            if contains_gift(item, target):
                return True
        elif item == target:
            return True
    return False
```

```python
print("toy:", contains_gift(box, "toy"))
print("watch:", contains_gift(box, "watch"))
print("phone:", contains_gift(box, "phone"))
print("Toy:", contains_gift(box, "Toy"))
print("book:", contains_gift([], "book"))
print("[[], ['pen']]:", contains_gift([[], ["pen"]], "pen"))
print("extra:", contains_gift([["a"], ["b"]], "b"))
```

**Exercise 3: Palindrome numbers**

```python
def isPalindromeNum(n):
    digits = str(n)
    def check(left, right):
        if left >= right:
            return True
        if digits[left] != digits[right]:
            return False
        return check(left + 1, right - 1)
    return check(0, len(digits) - 1)
```

```python
for n in [0, 7, 11, 123454321, 1357531, 168861]:
    print(f"isPalindromeNum({n}):", isPalindromeNum(n))
for n in [10, 112221]:
    print(f"isPalindromeNum({n}):", isPalindromeNum(n))
print("extra 12341:", isPalindromeNum(12341))
```

**Challenge Exercise 1: Unpack all gifts**

```python
def unpack(box):
    result = []
    for item in box:
        if isinstance(item, list):
            result += unpack(item)
        else:
            result.append(item)
    return result
print(unpack(box))
print(unpack([["ball", "ball"]]))
print(unpack([]))
```

**Challenge Exercise 2: Find the deepest box**

```python
def box_depth(box):
    deepest = 0
    for item in box:
        if isinstance(item, list):
            deepest = max(deepest, box_depth(item))
    return 1 + deepest
print(box_depth([]))
print(box_depth(["book", "pen"]))
print(box_depth([["book"]]))
print(box_depth(["book", [["toy"]]]))
print(box_depth([[]]))
```

**Testing and reflection**

## What stops each function?
- **count_gifts:** the for loop ends after the last item of a box. An empty box has no items, so it returns 0 without any recursive call (base case).
- **contains_gift:** it stops early with `True` when the target is found. If nothing matches, it stops after every item in every inner box has been checked and returns `False`.
- **isPalindromeNum:** `check()` stops when `left >= right` (all pairs matched, `True`) or when `digits[left] != digits[right]` (mismatch, `False`).

## What becomes smaller?
- **count_gifts and contains_gift:** each recursive call gets an inner box, which is one nesting level shallower than the box that holds it, so the problem shrinks until it reaches boxes with only gifts or no items.
- **isPalindromeNum:** `left` moves right and `right` moves left, so the part of the string still to be checked shrinks by 2 characters each call.

## Why can returning False too early fail?
- In `contains_gift`, an inner box that does not contain the target only means the target is not in that box.
- The target may still be in a later item or a later inner box.
- If the function returned `False` right after the first failed inner box, it would skip the rest and give a wrong answer.
- `False` must only be returned after all items have been checked.

## Extra test and the mistake it could detect

- **`isPalindromeNum(12341)` should be `False`:** the middle digits match but the outer ones do not, so it detects a function that forgets the mismatch case or returns `True` by default.
