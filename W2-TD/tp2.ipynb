**challenge excercise1**


```python
student_name = ("Alice", "Bob", "Charlie", "David")
print(student_name)
```

```python
students_set = set(student_name)
print("Set (duplicates removed):", students_set)
```

```python
search_name = "Charlie"
if search_name in students_set:
    print(f"{search_name} is in the set.")
else:
    print(f"{search_name} is not in the set.")
```

```python
action = input("Do you want to (add/remove) a name? ").strip().lower()
name = input("Enter the name: ").strip()
if action == "add":
    students_set.add(name)
    print(f"{name} added.")
elif action == "remove":
    students_set.discard(name) 
    print(f"{name} removed (if it existed).")
else:
    print("No valid action selected.")

print("Updated set:", students_set)

with open("students.txt", "a") as f:
    for student in students_set:
        f.write(student + "\n")
print("Set saved to students.txt")


loaded_set = set()
with open("students.txt", "r") as f:
    for line in f:
        loaded_set.add(line.strip())

print("Loaded from file:", loaded_set)


if loaded_set == students_set:
    print("Verification successful: loaded set matches the saved set.")
else:
    print("Verification failed: sets do not match.")
```

**Excercise2**

```python
import random
import time

def dice_roll(guess=(1, 3), n=100):
    start_time = time.time()          
    
    correct = 0
    for _ in range(n):
        roll = random.randint(1, 3)   
        if roll in guess:             
            correct += 1
    
    proportion = correct / n
    end_time = time.time()
    elapsed_time = end_time - start_time
    
    return proportion, elapsed_time



proportion, elapsed_time = dice_roll(guess=(1, 36), n=100000)
print(f"Proportion correctly guessed: {proportion}")
print(f"Running time: {elapsed_time:.6f} seconds")
```
