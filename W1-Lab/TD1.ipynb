## Task 6

```python
import wikipedia as wk
sum = wk.summary("Python (programming language)" )
print(sum)
```

```python
with open('wiki_python.txt', 'w') as f: 
    f.write(sum)
```

```python
def count_vowels_consonant_digit():
    vowels = 0
    consonants = 0
    digits = 0
    try:
        with open('wiki_python.txt', 'r') as f:
            content = f.read()
            for char in content:
                if char.isalpha():
                    if char.lower() in 'aeiou':
                        vowels += 1
                    else:
                        consonants += 1
                elif char.isdigit():
                    digits += 1
    except FileNotFoundError:
        print("File not found.")
    return vowels, consonants, digits

count_vowels_consonant_digit()
print(f"Vowels: {count_vowels_consonant_digit()[0]}, Consonants: {count_vowels_consonant_digit()[1]}, Digits: {count_vowels_consonant_digit()[2]}")

```
